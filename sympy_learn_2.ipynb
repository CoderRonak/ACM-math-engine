### System of Linear Equations
- linsolve([eq1, eq2, eq3], (x, y, z))

````python
eq1 = Eq(x + y + z - 1, 0)
eq1
````

````python
eq2 = Eq(x + y + 2*z - 3, 0)
eq2
````

````python
eq3 = Eq(x - y - z, 0)
````

````python
system = [eq1, eq2, eq3 ]
vars = [x, y, z]
````

````python
linsolve(system, vars)
````

### Solving Differential Equations

### Creating a function
To represent unknown function, we have to specify it is NOT a variable but a Function:

```
f = Symbol("f", cls = Function)
```
f(x) will be a function of x

### Solving a ODE

```
dsolve(diffeqn, func to solve for)
```




````python
f, g = symbols("f g", cls = Function)
````

````python
display(f(x), f(x).diff())
````

````python
f(g(x))
````

````python
f(g(x)).diff() # chain rule
````

````python
diffeq = Eq(f(x).diff(x, x) - 2*f(x).diff(x) + f(x), sin(x))
diffeq
````

````python
soln = dsolve(diffeq, f(x))
# gives f(x) = general solution
# gives an Eq() object
soln
````

````python
# To get RHS expression
soln = soln.rhs
soln
````

````python
# C1, C2 = symbols("C_1 C_2") -> NO NEED TO DEFINE CONSTANTS
````

````python
# particular solution
soln.subs({"C1":3, "C2":4}) # write in ""
````
