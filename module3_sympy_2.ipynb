## 2 variables, 1 constraint

```python
from sympy import *
```

```python
f, g, h = symbols("f g h", cls=Function)
```

```python
x, y, l, m = symbols("x y lambda mu")
```

```python
f = x*y
g = x**2/8 + y**2/2 - 1
```

```python
f_x, f_y, g_x, g_y = symbols("f_x f_y g_x g_y", cls=Function)
```

```python
f_x = f.diff(x)
f_y = f.diff(y)
g_x = g.diff(x)
g_y = g.diff(y)
```

```python
Eq(f_y, l*g_y)
```

```python
system = [
    Eq(f_x, l*g_x), Eq(f_y, l*g_y), Eq(g,0)
]
vars = (x, y, l)
```

```python
soln = solve(system, vars, dict=True)
soln
```

```python
pts = []
for pt in soln:
  pts.append((pt[x], pt[y]))

pts
```

```python
# Evaluate f at all pts
```

```python
evaluation = {}
```

```python
for pt in pts:
  xi, yi = pt[0], pt[1]
  val = f.subs({x:xi, y:yi})
  evaluation[(xi,yi)]=val

evaluation
```

```python
max_val = max(evaluation.values())
min_val = min(evaluation.values())
```

```python
list(evaluation.items())
```

```python
max_val, min_val
```

```python
maxima = [(a, b) for ((a, b), v) in evaluation.items() if v == max_val]
minima = [(a, b) for ((a, b), v) in evaluation.items() if v == min_val]

maxima
```
