```python
# 1. Evaluate the objective function f at each critical point
evaluated_solns = []
f_expr = funcs['f']['expr']

# Sort optimization variables for consistent coordinate ordering
sorted_op_vars = sorted(list(op_vars))

for soln in solns:
    # Substitute variables into our objective function
    val = f_expr.subs({all_vars[v]: soln[all_vars[v]] for v in sorted_op_vars if all_vars[v] in soln})
    evaluated_solns.append((soln, val))

# Find the overall maximum and minimum values
vals = [item[1] for item in evaluated_solns]
max_val = max(vals)
min_val = min(vals)

# 2. Build the structured table rows
rows = []
for soln, val in evaluated_solns:
    row = {}

    # Add Lagrange multiplier columns if they exist
    if 'l' in lagrange_var:
        row[f"${latex(lagrange_var['l'])}$"] = f"${latex(soln[lagrange_var['l']])}$"
    if 'm' in lagrange_var:
        row[f"${latex(lagrange_var['m'])}$"] = f"${latex(soln[lagrange_var['m']])}$"

    # Group optimization variables into a single tuple coordinate
    coord_str = f"({', '.join(sorted_op_vars)})"
    coord_val = tuple(soln[all_vars[v]] for v in sorted_op_vars)
    row[f"${coord_str}$"] = f"${latex(coord_val)}$"

    # Evaluated function column
    row[f"$f{coord_str}$"] = f"${latex(val)}$"

    # Conclusion column
    if val == max_val and val == min_val:
        conclusion = "Constant Value"
    elif val == max_val:
        conclusion = "**Maxima**"
    elif val == min_val:
        conclusion = "**Minima**"
    else:
        conclusion = "Saddle / Intermediate"
    row["Conclusion"] = conclusion

    rows.append(row)

# Create and display the Markdown table
df_result = pd.DataFrame(rows)
display(Markdown(df_result.to_markdown(index=False)))
```

3 variables, 2 constraints
- in this case
- f_x = lg_x + muh_x f_y = lg_y + muh_y f_z = lg_z + muh_z , g = 0 , h = 0 -> solve all these equations

User should be able to choose between the three types, enter the function, and the output should use latex to the fullest and look something like:

- Steps to optimise f = (func) subject to the constraints g = (func) & h = (func)

Solve the following equations simultaneously:
- delf = lamdadelg + mudelh, g=0, h=0
=> f_x = lg_x + muh_x eqn 1
-  f_y = lg_y + muh_y eqn 2
- f_z = lg_z + muh_z eqn 3
- g = 0 eqn 4
- h = 0 eqn 5

=> critical points (x,y,z) = (x1, y1,z1)... (actual solutions)

evaluating func at critical points
f|(x1,y1,z1) = val
....

therefore, func attains:
- maximum value at pts (x1, y1, z1) ..
- minimum value at pts (x2, y2, z2) ..

```python
from sympy import *
from sympy.parsing.sympy_parser import (
    convert_xor,
    implicit_multiplication_application,
    parse_expr,
    standard_transformations,
)
import re
import pandas as pd
from IPython.display import display, Latex, Markdown

# 1. Initialization and Setup
init_printing(use_latex='mathjax', order='grlex')
_TRANSFORMATIONS = standard_transformations + (
    implicit_multiplication_application,
    convert_xor,
)

# 2. User Input prompts
print("""1> 2 variables, 1 constraint
2> 3 variables, 1 constraint
3> 3 variables, 2 constraints""")

try:
    ch = int(input("Choose: "))
except ValueError:
    ch = 3

inps = {}
inps['f'] = input("Enter the function f: ")
inps['g'] = input("Enter the constraint g = 0: ")
if ch == 3:
    inps['h'] = input("Enter the constraint h = 0: ")

op_var_inp = input("Enter the optimisation variables: ")
op_vars = set(re.split(r'[;,\s]+', op_var_inp.strip()))

# 3. Parsing Functions and Generating Symbols
funcs = {}
func_list = ['f', 'g'] if ch in [1, 2] else ['f', 'g', 'h']
for i in func_list:
    s = Symbol(i)
    funcs[i] = {i: s, 'expr': parse_expr(inps[i].strip(), transformations=_TRANSFORMATIONS, local_dict={"e": E, "pi": pi})}

symbols_in_func = {}
for i, j in funcs.items():
    symbols_in_func[i] = set(sorted(map(lambda v: str(v), j['expr'].free_symbols)))

func_of = {}
for i, j in funcs.items():
    func_of[i] = symbols_in_func[i].intersection(op_vars)

func_of_display = {}
for func_sym, vars_sym in func_of.items():
    func_of_display[func_sym] = f"{func_sym}({', '.join(sorted(vars_sym))})"

# Display Parsed Functions
for func, func_display in func_of_display.items():
    display(Latex(fr"${func_display} = {latex(funcs[func]['expr'])} {'= 0' if func != 'f' else '' }$"))
    if func == "f":
        print("subject to the constraint(s): \n")

# Map symbols to SymPy Symbol objects
all_vars = {}
for func_sym, individual_func_symbols in symbols_in_func.items():
    for symbol in individual_func_symbols:
        if symbol not in all_vars:
            all_vars[symbol] = Symbol(symbol)

# 4. Computing Partial Derivatives
single_diff = {}
for i in funcs.keys():
    for j in op_vars:
        _ = f'{i}_{j}'
        s = Symbol(_)
        single_diff[_] = {f"{i}": i, f"{j}": all_vars[j], f"{s}": s, f"{s}_eval": funcs[i]['expr'].diff(all_vars[j])}

# Define Lagrange Multipliers
lagrange_var = {'l': Symbol('lambda')}
if ch == 3:
    lagrange_var['m'] = Symbol('mu')

# 5. Formulating the System of Equations
system_initial = []
system = []

for func_sym, func in funcs.items():
    if func_sym != 'f':
        system_initial.append(Eq(func[func_sym], 0))
        system.append(Eq(func['expr'], 0))

for single_diff_sym, single_diff_dict in single_diff.items():
    if single_diff_sym.startswith('g_'):
        v = single_diff_sym[-1]
        f_v_eval = single_diff[f'f_{v}'][f"f_{v}_eval"]
        lhs = f_v_eval

        g_v_eval = single_diff[f'g_{v}'][f"g_{v}_eval"]
        rhs = lagrange_var['l'] * g_v_eval

        if ch == 3:
            h_v_eval = single_diff[f'h_{v}'][f'h_{v}_eval']
            rhs = lagrange_var['l'] * g_v_eval + lagrange_var['m'] * h_v_eval

        system.append(Eq(lhs, rhs))

        # Symbolic representations
        f_v_sym = single_diff[f'f_{v}'][f"f_{v}"]
        lhs_sym = f_v_sym
        g_v_sym = single_diff[f'g_{v}'][f"g_{v}"]
        rhs_sym = lagrange_var['l'] * g_v_sym

        if ch == 3:
            h_v_sym = single_diff[f'h_{v}'][f'h_{v}']
            rhs_sym = lagrange_var['l'] * g_v_sym + lagrange_var['m'] * h_v_sym

        system_initial.append(Eq(lhs_sym, rhs_sym))

# Display the System
print("System of Lagrange Equations to Solve:")
display(system_initial)
print("In Explicit Expression Form:")
display(system)

# Determine variables to solve for
all_vars_to_solve_for = []
for v in all_vars.values():
    if f'{v}' in op_vars:
        all_vars_to_solve_for.append(v)
for v in lagrange_var.values():
    all_vars_to_solve_for.append(v)

# 6. Solving the Equations
solns = solve(system, all_vars_to_solve_for, dict=True, force=True)
print("Raw solutions:", solns)

# 7. Constructing the Results Table
if solns:
    evaluated_solns = []
    f_expr = funcs['f']['expr']
    sorted_op_vars = sorted(list(op_vars))

    for soln in solns:
        val = f_expr.subs({all_vars[v]: soln[all_vars[v]] for v in sorted_op_vars if all_vars[v] in soln})
        evaluated_solns.append((soln, val))

    # Classify extrema
    vals = [item[1] for item in evaluated_solns]
    max_val = max(vals)
    min_val = min(vals)

    rows = []
    for soln, val in evaluated_solns:
        row = {}
        if 'l' in lagrange_var:
            row[f"${latex(lagrange_var['l'])}$"] = f"${latex(soln.get(lagrange_var['l'], ''))}$"
        if 'm' in lagrange_var:
            row[f"${latex(lagrange_var['m'])}$"] = f"${latex(soln.get(lagrange_var['m'], ''))}$"

        coord_str = f"({', '.join(sorted_op_vars)})"
        coord_val = tuple(soln.get(all_vars[v], None) for v in sorted_op_vars)
        row[f"${coord_str}$"] = f"${latex(coord_val)}$"
        row[f"$f{coord_str}$"] = f"${latex(val)}$"

        if val == max_val and val == min_val:
            conclusion = "Constant Value"
        elif val == max_val:
            conclusion = "**Maxima**"
        elif val == min_val:
            conclusion = "**Minima**"
        else:
            conclusion = "Saddle / Intermediate"
        row["Conclusion"] = conclusion
        rows.append(row)

    df_result = pd.DataFrame(rows)
    display(Markdown("### Final Critical Points and Extrema Evaluation"))
    display(Markdown(df_result.to_markdown(index=False)))
else:
    print("No solutions found.")
```
