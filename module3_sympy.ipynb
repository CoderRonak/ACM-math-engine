Determine all critical points of $$f(x,y) = x^{3} - 3x + y^{2}$$

```python
f, f_x, f_y = symbols("f f_x f_y", cls=Function)
x, y = symbols("x y")
```

```python
f = x**3 - 3*x + y**2
f
```

Step 1:
Find $f_x = \frac{∂f}{∂x}$ & $f_y = \frac{∂f}{∂y}$

```python
f_x = f.diff(x)
f_x
```

```python
f_y = f.diff(y)
f_y
```

Step 2: Equate $f_x = 0$ & $f_y = 0$ to get critical points

```python
x_soln = solveset(f_x)
x_soln
```

```python
y_soln = solveset(f_y)
y_soln
```

```python
# Critical points:
pts = set()
for x_i in x_soln:
  for y_i in y_soln:
    pts.add((x_i,y_i))

pts
```

Critical points could be local maxima, minima or saddle

Step 3: Evaluate $det(H(f))$ at critical points where $H(f) = \left[\begin{matrix}f_{x x} & f_{x y}\\f_{y x} & f_{y y}\end{matrix}\right]$

```python
f_x_x, f_x_y, f_y_x, f_y_y = symbols("f_x_x f_x_y f_y_x f_y_y")
```

Find $f_{xx}, f_{xy} ,f_{yx}, f_{yy}$

```python
f_x_x = f_x.diff(x)
f_y_y = f_y.diff(y)
f_y_x = f_x_y = f_x.diff(y)
```

Find $det(H(f))$

```python
H = Matrix([[f_x_x, f_x_y],[f_y_x, f_y_y]])
H_det = H.det()
H_det
```

Evaluate $det(H(f))$ at critical points

```python
D_f = {}
for pt in pts:
  D_f[pt]= (H_det.subs({x: pt[0], y:pt[1]}))

D_f
```

```python
f_x_x
```

```python
for pt, D_at_pt in D_f.items():
  if D_at_pt > 0:
    f_x_x_at_pt = f_x_x.subs({x: pt[0], y:pt[1]})
    print(f_x_x_at_pt)
    if f_x_x_at_pt > 0:
      print(pt, "is Local Minima")
    else:
      print(pt, "is Local Maxima")
  elif D_at_pt < 0:
    print(pt, "is Saddle Point")
  else:
    print("Inconclusive Test")
```

# ENTIRE BEAUTIFUL PROGRAM IN THE BELOW CELL

```python
from IPython.display import display, Latex, Markdown
from sympy import *
from sympy.parsing.sympy_parser import (
    convert_xor,
    implicit_multiplication_application,
    parse_expr,
    standard_transformations,
)

init_printing(use_latex='mathjax')


_TRANSFORMATIONS = standard_transformations + (
    implicit_multiplication_application,
    convert_xor,
)

f_inp = input("Enter the function: ")

f = parse_expr(f_inp.strip(), transformations=_TRANSFORMATIONS, local_dict={"e": E, "pi": pi})
x, y, z = symbols("x y z")

display(Markdown(fr"### <center> *Steps to find extremes of $f = {latex(f)}$*"))

no_of_var = len(f.free_symbols)
if no_of_var == 1:
    (x,) = f.free_symbols  # 'x'
    x = Symbol(str(x))
elif no_of_var > 1:
    s = []
    for symbol in sorted(f.free_symbols, key=str):
        s.append(str(symbol))

    s = " ".join(s)

    if no_of_var == 2:
        x, y = symbols(s)
    elif no_of_var == 3:
        x, y, z = symbols(s)

f_x = Symbol(f"f_{x}")
f_y = Symbol(f"f_{y}")

# display(f, f_x, f_y)

f_x = f.diff(x)
f_y = f.diff(y)
# display(Latex(fr"$f_{{{x}}} = {latex(f_x)}$"))
# display(Latex(fr"$f_{{{y}}} = {latex(f_y)}$"))

# Solve the two derivative equations as one system. Solving each equation
# independently can pair coordinates that do not satisfy both equations.
critical_solutions = solve((f_x, f_y), (x, y), dict=True)
parametric_solutions = [
    solution for solution in critical_solutions
    if x not in solution or y not in solution
]
pts = {
    (solution[x], solution[y])
    for solution in critical_solutions
    if x in solution and y in solution
}

# 1. Print the first derivatives side-by-side
display(Latex(fr"$f_{{{x}}} = {latex(f_x)} \text{{  and  }} f_{{{y}}} = {latex(f_y)}$"))
print()

# 2. Print the text header for system of equations
print(f"For critical points, f{x} = 0 & f{y} = 0")
print()

# 3. Print the system implication line with the solved coordinate pairs
if parametric_solutions:
    solution_text = ", ".join(latex(solution) for solution in parametric_solutions)
    display(Latex(fr"$f_{{{x}}} = 0 \ \& \ f_{{{y}}} = 0 \implies \text{{critical solution set includes }} {solution_text}$"))
else:
    formatted_solution = ", ".join(
        f"({latex(pt_x)}, {latex(pt_y)})" for pt_x, pt_y in sorted(pts, key=str)
    )
    display(Latex(fr"$f_{{{x}}} = 0 \ \& \ f_{{{y}}} = 0 \implies (x, y) = {formatted_solution}$"))

print()
print("Therefore, critical points are: ")

# 4. Print the critical points
formatted_pts = " & ".join([f"({pt[0]},{pt[1]})" for pt in sorted(pts, key=str)])
if parametric_solutions:
    print("described by", solution_text)
elif pts:
    print(formatted_pts)
else:
    print("none")

print()
print()

f_x_x, f_x_y, f_y_x, f_y_y = symbols(f"f_{x}_{x} f_{x}_{y} f_{y}_{x} f_{y}_{y}")

f_x_x = f_x.diff(x)
f_y_y = f_y.diff(y)
f_y_x = f_x_y = f_x.diff(y)

H = Matrix([[f_x_x, f_x_y],[f_y_x, f_y_y]])
display(Latex(fr"""$H(f) =
\left[\begin{{matrix}}
    f_{{{x}{x}}} & f_{{{x}{y}}} \\
    f_{{{y}{x}}} & f_{{{y}{y}}}
\end{{matrix}}\right] = {latex(H)}$"""))
print()

H_det = H.det()
display(Latex(fr"$det(H(f)) = {latex(H_det)}$"))
print()
print()

D_f = {}
for pt in pts:
  D_f[pt]= (H_det.subs({x: pt[0], y:pt[1]}))


for pt, D_at_pt in D_f.items():
    # Extract coordinates cleanly from your point set/tuple
    # Using the sequence unpacking technique discussed earlier
    pt_x, pt_y = pt

    # 1. Evaluate the mathematical conditions
    if D_at_pt > 0:
        f_x_x_at_pt = f_x_x.subs({x: pt_x, y: pt_y})
        f_xx_latex = latex(f_x_x_at_pt)

        if f_x_x_at_pt > 0:
            conclusion = fr"> 0 \ \& \ f_{{{x}{x}}} = {f_xx_latex} > 0 \implies ({pt_x}, {pt_y}) \text{{ is a Local Minima}}"
        else:
            conclusion = fr"> 0 \ \& \ f_{{{x}{x}}} = {f_xx_latex} < 0 \implies ({pt_x}, {pt_y}) \text{{ is a Local Maxima}}"

    elif D_at_pt < 0:
        conclusion = fr"< 0 \implies ({pt_x}, {pt_y}) \text{{ is a Saddle Point}}"

    else:
        conclusion = fr"= 0 \implies \text{{The Second Derivative Test is Inconclusive}}"

    # 2. Combine the evaluation and the conclusion into one unified LaTeX block
    # \mid represents the evaluation bar "|" and \implies renders the beautiful "=>" arrow
    final_latex = fr"$\det(H(f))\mid_{{({pt_x}, {pt_y})}} = {latex(D_at_pt)} {conclusion}$"

    # 3. Output the beautifully typeset line
    display(Latex(final_latex))
```
