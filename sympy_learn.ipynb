### Continued Fraction

```python
def list_to_frac(l):
    expr = sympy.Integer(0) # sympy Integer object
    for i in reversed(l[1:]):
        expr += i
        expr = 1/expr
    return l[0] + expr
```

```python
list_to_frac([x, y, z, p, q])
```
